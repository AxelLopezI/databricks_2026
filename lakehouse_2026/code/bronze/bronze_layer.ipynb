# Read from CSV and write silver table

```python
for i in INGESTION_CONFIG: 
    print(f"Ingesting {i["name"]}.{i["extension"]} in table {i["table"]}")
    
    # Define the volume path.
    path = f"/Volumes/workspace/bronze/source_system/{i["folder"]}/{i["name"]}.{i["extension"]}"

    # Read the CSV file into a Spark DataFrame.
    df = (
        spark.read
            .format(f"{i["extension"]}")
            .option("header", "true")
            .option("inferSchema", "true")
            .load(path)
    )

    # Write DataFrame as a Delta table
    (
        df.write
            .mode("overwrite")
            .format("delta")
            .saveAsTable(f"bronze.{i["table"]}")
    )
```
