# Join between silver tables

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY cp.start_date, cp.product_number) AS product_key, 
    cp.product_id,
    cp.product_number,
    cp.product_name,
    cp.category_id,
    epc.category,
    epc.subcategory,
    epc.maintenance_flag,
    cp.product_line,
    cp.start_date
FROM silver.crm_products cp
LEFT JOIN silver.erp_product_category epc
    ON cp.category_id = epc.category_id
"""

df = spark.sql(query)
df.display()
```

# Writing gold table

```python
(
    df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable("workspace.gold.dim_products")
)
```
